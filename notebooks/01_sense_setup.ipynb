## Notebook 01: SENSE Setup

This notebook runs the SENSE pipeline on all background databases to add social flows from PSILCA. It also runs SENSE for the two validation projects (arithmetic and geometric averaging).

**What it does:**
1. Runs `run_sense()` on the base ecoinvent database (`ecoinvent-3.10.1-cutoff`).
2. Runs `run_sense()` on the regionalized ecoinvent database (`ecoinvent-3.10.1-cutoff regionalized`).
3. Runs `run_sense()` on the Regioinvent database.
4. Runs `run_sense()` twice for validation (ecoinvent 3.11, arithmetic and geometric mean).

Each run parses PSILCA data, matches ecoinvent activities to PSILCA sectors via semantic embeddings, averages social indicators with price weighting, and writes social biosphere flows and impact methods into the Brightway database.

**Inputs:**
- Brightway projects from notebook 00.
- PSILCA v4 EcoSpold01 export (`Data/PSILCA/EcoSpold01/`).
- Ecoinvent EcoSpold02 datasets (`Data/ecoinvent/.../datasets/`).
- Ecoinvent geography overview Excel (`Data/Database-Overview-for-ecoinvent-v3.10_29.04.24.xlsx`).

**Outputs:**
- Social biosphere flows and LCIA impact methods added to each Brightway database.
- Intermediate files (embeddings, matched JSONs) in `notebooks/Data/Output/`.
