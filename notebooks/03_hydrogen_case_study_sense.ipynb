## Notebook 03: Hydrogen Case Study SENSE

This notebook runs the SENSE pipeline on the foreground hydrogen databases to add social flows to the case study inventories.

**What it does:**
1. Runs `run_sense()` on the regionalized hydrogen inventory database.
2. Runs `run_sense()` on the regionalized hydrogen subset database.
3. Runs `run_sense()` on the regionalized methanol inventory database.

Each run operates in foreground mode, estimating prices from background database inputs and matching activities to PSILCA sectors.

**Inputs:**
- Brightway project with foreground databases from notebook 02.
- PSILCA v4 EcoSpold01 export.
- Ecoinvent EcoSpold02 datasets.
- Ecoinvent geography overview Excel.

**Outputs:**
- Social biosphere flows and LCIA impact methods added to each foreground database in Brightway.

```python
from pathlib import Path
from sense import run_sense, SenseConfig
```

```python
### Run SENSE on hydrogen database

results = run_sense(
    project="sense_project",
    technosphere_db="01_hydrogen_inventories_regionalized",
    biosphere_db="biosphere3_spatialized_flows",
    foreground_active=True,
    foreground_database="01_hydrogen_inventories_regionalized",
    background_databases=["ecoinvent-3.10.1-cutoff", "ecoinvent-3.10.1-cutoff regionalized", "Regioinvent"],
    ecoinvent_dir=Path("../Data/ecoinvent/ecoinvent 3.10.1_cutoff_ecoSpold02/datasets"),
    psilca_dir=Path("../Data/PSILCA/EcoSpold01"),
    ecoinvent_version="3.10.1",
    psilca_version="4.0",
    config=SenseConfig(),
    ecoinvent_geographies_excel=Path("../Data/Database-Overview-for-ecoinvent-v3.10_29.04.24.xlsx"),
 )
```
